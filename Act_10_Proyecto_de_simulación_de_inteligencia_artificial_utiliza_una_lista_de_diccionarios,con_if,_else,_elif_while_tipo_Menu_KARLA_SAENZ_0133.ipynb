```python
# --- SIMULADOR DE IA: MANAGER DE INFLUENCERS ---

# Se inicializa la lista vacía y un diccionario vacío como base [4-6]
influencer = []
losinfluencers = {}

print("🤖 BIENVENIDO AL SISTEMA 'AI-TALENT' 🤖")

# Iniciamos un ciclo infinito para el menú interactivo [7, 8]
while True:
    print("\n--- MENÚ DE CONTROL ---")
    print("1. Registrar Influencer (Añadir a diccionario)")
    print("2. Mostrar Comunidad (Ciclo for)")
    print("3. Consultar IA: Informe de Popularidad")
    print("4. Eliminar último registro")
    print("5. Salir")

    opcion = input("\nSelecciona una opción (1-5): ")

    # Estructura selectiva para manejar las opciones del menú [9, 10]
    if opcion == "1":
        # Creamos un nuevo diccionario vacío para los datos actuales [6]
        losinfluencers = {}

        # Entrada de datos por parte del usuario [11, 12]
        nombre = input("Nombre del Influencer: ")
        fans = int(input("Cantidad de seguidores (solo números): "))

        # Almacenamos los datos en el diccionario usando claves [6, 13]
        losinfluencers["nombre"] = nombre
        losinfluencers["seguidores"] = fans

        # Agregamos el diccionario a nuestra lista principal [5, 14]
        influencer.append(losinfluencers)
        print(f"✅ IA: '{nombre}' ha sido registrado en la base de datos.")

    elif opcion == "2":
        # Verificamos si la lista tiene elementos con if/else [9, 15]
        if len(influencer) == 0:
            print("❌ IA: No hay influencers registrados todavía.")
        else:
            print("\n--- LISTA DE TALENTOS ---")
            # Utilizamos un ciclo for para recorrer la lista de diccionarios [7, 16]
            for i in influencer:
                print(f"🌟 Perfil: {i['nombre']} | 👥 Seguidores: {i['seguidores']}")

    elif opcion == "3":
        if len(influencer) == 0:
            print("❌ IA: Sin datos para analizar.")
        else:
            print("\n--- ANÁLISIS DE TENDENCIAS (IA) ---")
            for i in influencer:
                # La IA toma decisiones basadas en condiciones lógicas [9, 17]
                if i["seguidores"] >= 1000000:
                    categoria = "Nivel: DIAMANTE 💎 (Leyenda)"
                elif i["seguidores"] >= 100000:
                    categoria = "Nivel: ORO 🥇 (En tendencia)"
                else:
                    categoria = "Nivel: BRONCE 🥉 (Creciendo)"

                print(f"Análisis de {i['nombre']}: {categoria}")

    elif opcion == "4":
        if len(influencer) > 0:
            # Eliminamos el último diccionario de la lista [18, 19]
            borrado = influencer.pop()
            print(f"🗑️ IA: Se ha eliminado a '{borrado['nombre']}'.")
        else:
            print("❌ IA: La lista ya está vacía.")

    elif opcion == "5":
        print("👋 IA: Cerrando sistema... ¡Suerte con tus redes!")
        # Rompemos el ciclo while True para finalizar el programa [8, 20]
        break

    else:
        # Respuesta en caso de que la opción no sea válida [9, 10]
        print("⚠️ IA: Error. Por favor elige una opción del 1 al 5.")
```
