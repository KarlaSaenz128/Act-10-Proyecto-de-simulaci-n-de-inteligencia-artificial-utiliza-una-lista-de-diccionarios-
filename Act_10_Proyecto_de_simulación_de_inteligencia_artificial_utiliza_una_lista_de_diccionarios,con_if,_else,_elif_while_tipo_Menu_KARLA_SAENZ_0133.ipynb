```python
# --- SIMULACIÓN DE IA: MANAGER DE INFLUENCERS ---

# Al inicio, la lista de influencers está vacía
influencers = []

print("✨ BIENVENIDO A 'AI-FLUENCE' - TU MANAGER VIRTUAL ✨")

# Ciclo infinito para el menú interactivo
while True:
    print("\n--- MENÚ DE GESTIÓN ---")
    print("1. Registrar nuevo Influencer")
    print("2. Ver mi comunidad (Lista de diccionarios)")
    print("3. Consultar IA: Análisis de Tendencias")
    print("4. Eliminar último registro")
    print("5. Salir del sistema")

    opcion = input("\nSelecciona una opción (1-5): ")

    if opcion == "1":
        # Se inicializa un diccionario vacío para el nuevo registro
        datos_influencer = {}

        # Entrada de datos
        nombre = input("Ingresa el nombre del influencer: ")
        seguidores = int(input("¿Cuántos seguidores tiene? (solo números): "))

        # Se llena el diccionario con la información
        datos_influencer["nombre"] = nombre
        datos_influencer["seguidores"] = seguidores

        # Se agrega el diccionario a la lista principal
        influencers.append(datos_influencer)
        print(f"🤖 IA: '{nombre}' ha sido añadido a tu base de datos.")

    elif opcion == "2":
        if len(influencers) == 0:
            print("🤖 IA: Tu lista está vacía. ¡Necesitas buscar talento!")
        else:
            print("\n--- LISTA ACTUAL DE INFLUENCERS ---")
            # Ciclo for para recorrer la lista de diccionarios
            for i in influencers:
                print(f"🌟 Nombre: {i['nombre']} | 👥 Seguidores: {i['seguidores']}")

    elif opcion == "3":
        if len(influencers) == 0:
            print("🤖 IA: No hay datos suficientes para analizar.")
        else:
            print("\n--- INFORME DE LA IA ---")
            for i in influencers:
                # La IA analiza los datos usando if, elif y else
                if i["seguidores"] > 1000000:
                    status = "Nivel: SUPER ESTRELLA 💎"
                elif i["seguidores"] >= 100000:
                    status = "Nivel: TENDENCIA 🚀"
                else:
                    status = "Nivel: EN CRECIMIENTO 🌱"

                print(f"Análisis de {i['nombre']}: {status}")

    elif opcion == "4":
        if len(influencers) > 0:
            # Elimina el último diccionario de la lista
            eliminado = influencers.pop()
            print(f"🤖 IA: Se ha eliminado a '{eliminado['nombre']}'.")
        else:
            print("🤖 IA: Nada que eliminar.")

    elif opcion == "5":
        print("🤖 IA: Guardando datos... Sesión cerrada. ¡Adiós!")
        break # Rompe el ciclo while True para salir

    else:
        print("🤖 IA: Error 404 - Opción no válida. Intenta de nuevo.")
        print("elaborado por karla saenz 0133")
```
